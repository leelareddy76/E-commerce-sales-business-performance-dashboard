# Interactive E‑commerce Analysis

This notebook loads the synthetic dataset, shows key KPIs, and renders charts for monthly trends, top products, and revenue by location. Use the kernel from the project's virtual environment (.venv).

```python
import sys
from pathlib import Path
# ensure project root is importable when the notebook runs from notebooks/
sys.path.insert(0, str(Path.cwd()))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.analytics import load_data, compute_kpis, top_products, revenue_by_location, monthly_trends

sns.set(style='whitegrid')
```

```python
# Load data (adjust path if you placed CSVs elsewhere)
data_dir = 'data'
ds = load_data(data_dir)
orders = ds['orders']
products = ds['products']
locations = ds['locations']

# Compute KPIs and display
kpis = compute_kpis(orders)
kpis_series = pd.Series(kpis)
display(kpis_series.to_frame('value'))
```

```python
# Monthly revenue trend
months = monthly_trends(orders)
plt.figure(figsize=(10,5))
sns.lineplot(data=months, x='month', y='revenue', marker='o')
plt.title('Monthly Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Top 10 products by revenue
top = top_products(orders, products, top_n=10)
plt.figure(figsize=(10,6))
sns.barplot(data=top, x='revenue', y='product_name', palette='viridis')
plt.title('Top 10 Products by Revenue')
plt.tight_layout()
plt.show()
```

```python
# Revenue by location
rloc = revenue_by_location(orders, locations)
plt.figure(figsize=(8,5))
sns.barplot(data=rloc, x='revenue', y='location_name', palette='magma')
plt.title('Revenue by Location')
plt.tight_layout()
plt.show()
```
